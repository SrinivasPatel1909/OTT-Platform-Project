```python
import pandas as pd

# Merge Viewing Activity with User Profile
ott_master = pd.merge(
    viewing_activity,
    user_profile,
    on="User_ID",
    how="inner"
)

# Merge Subscription Retention
ott_master = pd.merge(
    ott_master,
    subscription_retention,
    on="User_ID",
    how="left"
)

# Merge Ratings Feedback
ott_master = pd.merge(
    ott_master,
    ratings_feedback,
    on=["User_ID", "Content_ID"],
    how="left"
)

# Merge Content Library
ott_master = pd.merge(
    ott_master,
    content_library,
    left_on="Content_ID",
    right_on="show_id",
    how="left"
)

print("Master Dataset Created Successfully")
print(ott_master.shape)

ott_master.head()
```


# **Task 1** :- Analyze User Engagement Across Different Customer Segments



```python
# Analysis 1 : Which subscription plan has the highest average watch duration?

import pandas as pd
import numpy as np

# Removed: ott_master = pd.merge(user_profile, viewing_activity, on='User_ID', how='inner')
# Assuming ott_master from cell C9uvt1T-6TIa is the correct one to use for all analyses.

age_summary = (
    ott_master
    .groupby("Age_Group")["Watch_Duration_Minutes"]
    .mean()
    .reset_index()
)

highest_age = age_summary.loc[
    age_summary["Watch_Duration_Minutes"].idxmax()
]

print(age_summary)

print("\nAge Group with Highest Engagement")
print(highest_age)
```

```python

# Analysis 2 : Which subscription plan has the highest average content completion percentage?


completion = (
    ott_master.groupby("Subscription_Type")["Completion_Percentage"]
    .mean()
    .round(2)
    .reset_index()
)

highest_completion = completion.loc[
    completion["Completion_Percentage"].idxmax()
]

print("Average Completion Percentage\n")
print(completion)

print("\nHighest Completion Percentage")
print(highest_completion)
```

```python

# Analysis 3 :- Which subscription plan receives the highest average user rating?


import pandas as pd

rating_analysis = pd.merge(
    user_profile[["User_ID", "Subscription_Type"]],
    ratings_feedback[["User_ID", "Rating"]],
    on="User_ID",
    how="inner"
)

rating_summary = (
    rating_analysis
    .groupby("Subscription_Type")["Rating"]
    .mean()
    .round(2)
    .reset_index()
)

highest_rating = rating_summary.loc[
    rating_summary["Rating"].idxmax()
]

print("Average Rating by Subscription Plan\n")
print(rating_summary)

print("\nSubscription Plan with Highest Average Rating")
print(highest_rating)
```

Task 1 Interpretation :- The analysis shows that users aged 36–50 are the most engaged, the Premium plan achieves the highest content completion, and the Basic plan receives the highest ratings, supporting better subscription and customer engagement strategies.

**Task 2: Analyze Overall User Viewing Engagement**

```python

# Analysis 1 : Which user engagement level records the highest average watch duration?

import pandas as pd

engagement_data = pd.merge(
    user_profile[["User_ID", "Engagement_Level"]],
    viewing_activity[["User_ID", "Watch_Duration_Minutes"]],
    on="User_ID",
    how="inner"
)

engagement_summary = (
    engagement_data
    .groupby("Engagement_Level")["Watch_Duration_Minutes"]
    .mean()
    .round(2)
    .reset_index()
)

highest_engagement = engagement_summary.loc[
    engagement_summary["Watch_Duration_Minutes"].idxmax()
]

print("Average Watch Duration by Engagement Level\n")
print(engagement_summary)

print("\nHighest User Engagement Level")
print(highest_engagement)
```

```python
# Analysis 2 : Which device type is most preferred by highly engaged users?

high_users = user_profile[
    user_profile["Engagement_Level"] == "High"
]

device_summary = (
    high_users["Device_Type"]
    .value_counts()
    .reset_index()
)

device_summary.columns = ["Device_Type", "Total_Users"]

print(device_summary)

print("\nMost Preferred Device")

print(device_summary.iloc[0])
```

```python

# Analysis 3 : Which time of day records the highest average watch duration?


time_summary = (
    viewing_activity
    .groupby("Time_of_Day")["Watch_Duration_Minutes"]
    .mean()
    .round(2)
    .reset_index()
)

highest_time = time_summary.loc[
    time_summary["Watch_Duration_Minutes"].idxmax()
]

print("Average Watch Duration by Time of Day\n")
print(time_summary)

print("\nPeak Viewing Time")
print(highest_time)
```

Task 2 Interpretation : The findings indicate that night is the peak viewing period, mobile is the most preferred device, and engagement patterns vary across users, enabling the platform to improve content scheduling and customer engagement strategies.

```python

# Analysis 3 : Which gender spends more time watching content?

gender_watch = (
    ott_master
    .groupby("Gender")["Watch_Duration_Minutes"]
    .mean()
    .round(2)
    .reset_index()
)

highest_gender = gender_watch.loc[
    gender_watch["Watch_Duration_Minutes"].idxmax()
]

print(gender_watch)

print("\nGender with Highest Watch Duration")
print(highest_gender)
```

**Task 4 :- Analyze User Content Consumption**

```python
# Analysis 1 : Which content receives the highest average rating?


content_rating = (
    ratings_feedback
    .groupby("Content_ID")["Rating"]
    .mean()
    .round(2)
    .reset_index()
)

highest_content = content_rating.loc[
    content_rating["Rating"].idxmax()
]

print(content_rating)

print("\nHighest Rated Content")
print(highest_content)
```

```python

# Analysis 2 : Which content receives the highest number of likes?


likes = (
    ratings_feedback[ratings_feedback["Liked_Flag"]=="Y"]
    .groupby("Content_ID")
    .size()
    .reset_index(name="Total_Likes")
)

highest_like = likes.loc[
    likes["Total_Likes"].idxmax()
]

print(likes)

print("\nMost Liked Content")
print(highest_like)
```

```python
# Analysis 3 : Which feedback category is most common?


feedback = (
    ratings_feedback["Feedback_Category"]
    .value_counts()
    .reset_index()
)

feedback.columns=["Feedback_Category","Total"]

print(feedback)

print("\nMost Common Feedback")
print(feedback.iloc[0])
```

**Task 4 Interpretation :-  The platform should promote highly rated and well-received content while using positive user feedback to strengthen customer satisfaction, increase engagement, and improve content investment decisions.**
